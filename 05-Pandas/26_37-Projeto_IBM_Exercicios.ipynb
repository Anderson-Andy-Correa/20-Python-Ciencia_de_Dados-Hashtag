## O dataset

O dataset a ser utilizado é o *IBM HR Analytics Employee Attrition & Performance*, disponível no Kaggle. O dataset contém informações sobre funcionários de uma empresa fictícia, como idade, gênero, estado civil, nível de satisfação, entre outros. O objetivo é prever a probabilidade de um funcionário sair da empresa.

[Link do Kaggle](https://www.kaggle.com/datasets/pavansubhasht/ibm-hr-analytics-attrition-dataset/data)

Os termos *attrition* ("atrito") e *turnover* ("rotatividade") são frequentemente usados de forma intercambiável no contexto de recursos humanos e gestão de força de trabalho, mas eles se referem a conceitos ligeiramente diferentes em relação às mudanças no número e na composição dos empregados dentro de uma organização. Compreender a distinção entre esses dois é importante para analisar e relatar com precisão a dinâmica da força de trabalho.

### Atrito

Atrito refere-se à redução gradual de empregados por meio de circunstâncias naturais, como aposentadoria, demissão voluntária por motivos pessoais, ou a decisão de não substituir empregados que estão de saída. Atrito pode levar a uma diminuição no tamanho da força de trabalho, mas é caracterizado por sua natureza voluntária e muitas vezes incontrolável. Organizações podem permitir o atrito para reduzir o tamanho de sua força de trabalho sem recorrer a demissões, vendo isso como uma forma menos disruptiva de ajustar seus níveis de pessoal.

Em análises, quando você está analisando um conjunto de dados para atrito, você frequentemente olha para as razões por trás das saídas voluntárias e a taxa na qual estas ocorrem naturalmente ao longo do tempo. Taxas de atrito podem fornecer insights sobre a satisfação dos empregados e tendências de longo prazo da força de trabalho.

### Rotatividade

Rotatividade, por outro lado, engloba um escopo mais amplo de partidas de empregados, incluindo tanto renúncias voluntárias quanto demissões involuntárias (como demissões ou términos por justa causa). Porém, ocorre a substituição da força de trabalho. Rotatividade é uma medida de quantos empregados estão deixando a companhia por qualquer motivo. Uma alta taxa de rotatividade pode indicar problemas com satisfação no trabalho, cultura do local de trabalho, ou estabilidade organizacional, e frequentemente requer ação da gestão para endereçar as causas subjacentes.

Rotatividade é calculada dividindo o número de empregados que saem pelo número médio de empregados, geralmente em uma base anual, e expressando isso como uma porcentagem. Esta métrica é crucial para entender como as saídas de empregados, tanto voluntárias quanto involuntárias, afetam a capacidade da organização de manter uma força de trabalho estável e eficaz.

### Diferenças Chave

- **Natureza**: Atrito geralmente é voluntário e pode ser devido a fatores naturais ou incontroláveis, enquanto rotatividade inclui tanto saídas voluntárias quanto demissões involuntárias.
- **Implicações para Gestão**: Atrito pode ser parte de uma estratégia de ajuste natural da força de trabalho e nem sempre é negativo, enquanto alta rotatividade é frequentemente um sinal de problemas dentro da organização que precisam ser abordados.
- **Foco Estratégico**: Análise de atrito pode focar em entender tendências de longo prazo da força de trabalho e planejamento de aposentadoria, enquanto análise de rotatividade é frequentemente usada para identificar questões imediatas com satisfação de empregados, engajamento, ou práticas de gestão que podem estar impactando a retenção de staff.

Enquanto ambos atrito e rotatividade lidam com empregados deixando a organização, eles diferem em seu escopo e as implicações estratégicas para gerenciar e entender a dinâmica da força de trabalho. No seu conjunto de dados de empregados, distinguir entre atrito e rotatividade pode ajudar na customização de estratégias para planejamento da força de trabalho, retenção e desenvolvimento organizacional.

```python
import pandas as pd

DADOS = r".\Dados\employee_attrition.csv"

df = pd.read_csv(DADOS)

display(df)
```

## Exercícios

- remover colunas com valores únicos
- encontrar os registros dos 10 funcionários mais velhos e dos 10 mais novos
- encontrar os registros dos 10 funcionários com mais tempo de empresa e dos 10 com menos tempo
- encontrar os registros dos 10 funcionários com mais tempo no cargo (sem promoção) e dos 10 com menos tempo
- agrupamentos (em cada caso, mostrar também o agrupamento por saída ou não do funcionário)
  - contagem de funcionários por departamento
  - média de salário por departamento
  - gênero por departamento
  - gênero por cargo
  - média de salário por cargo
  - média de salário por cargo e nível
  - contagem de gênero por departamento
  - média de salário por nível educacional

```python
# Remover colunas com valores únicos
# Remoção da coluna EmployeeNumber por conta de não agregar à análise da base de dados
remove_single_value_columns = [column for column, value in df.nunique().items() if value == 1]
df_colunas_removidas = df.drop(columns=remove_single_value_columns + ["EmployeeNumber"])
display(df_colunas_removidas)
```

```python
# Encontrar os registros dos 10 funcionários mais velhos e dos 10 mais novos
df_funcionarios_mais_novos_e_mais_velhos = pd.concat(
    [df_colunas_removidas.sort_values(by="Age").head(10),
      df_colunas_removidas.sort_values(by="Age").tail(10)]
    )
display(df_funcionarios_mais_novos_e_mais_velhos)

df_funcionarios_mais_novos = df_colunas_removidas.nsmallest(10, "Age")
display(df_funcionarios_mais_novos)

df_funcionarios_mais_velhos = df_colunas_removidas.nlargest(10, "Age")
display(df_funcionarios_mais_velhos)
```

```python
# Encontrar os registros dos 10 funcionários com mais tempo de empresa e dos 10 com menos tempo
df_funcionarios_mais_tempo_empresa_e_menos_tempo_empresa = pd.concat(
    [df_colunas_removidas.sort_values(by="TotalWorkingYears").head(10),
      df_colunas_removidas.sort_values(by="TotalWorkingYears").tail(10)]
    )
display(df_funcionarios_mais_tempo_empresa_e_menos_tempo_empresa)

df_funcionarios_mais_tempo_empresa = df_colunas_removidas.nlargest(10, "TotalWorkingYears")
display(df_funcionarios_mais_tempo_empresa)

df_funcionarios_menos_tempo_empresa = df_colunas_removidas.nsmallest(10, "TotalWorkingYears")
display(df_funcionarios_menos_tempo_empresa)
```

```python
# Encontrar os registros dos 10 funcionários com mais tempo no cargo (sem promoção) e dos 10 com menos tempo
df_funcionarios_mais_tempo_cargo_e_menos_tempo_cargo = pd.concat(
    [df_colunas_removidas.sort_values(by="YearsSinceLastPromotion").head(10),
      df_colunas_removidas.sort_values(by="YearsSinceLastPromotion").tail(10)]
    )
display(df_funcionarios_mais_tempo_cargo_e_menos_tempo_cargo)

df_funcionarios_mais_tempo_cargo = df_colunas_removidas.nlargest(10, "YearsSinceLastPromotion")
display(df_funcionarios_mais_tempo_cargo)

df_funcionarios_menos_tempo_cargo = df_colunas_removidas.nsmallest(10, "YearsSinceLastPromotion")
display(df_funcionarios_menos_tempo_cargo)
```

```python
# Analisando as situações de pessoas que tem maior tempo sem aumento
display(df_colunas_removidas.nlargest(10, "YearsSinceLastPromotion")[["Age", "JobRole", "YearsSinceLastPromotion", "JobLevel", "YearsAtCompany"]])

# Analisando as situações de pessoas que tem menos tempo sem aumento
display(df_colunas_removidas.nsmallest(10, "YearsSinceLastPromotion")[["Age", "JobRole", "YearsSinceLastPromotion", "JobLevel", "YearsAtCompany"]])
```

```python
colunas_categoricas_nao_ordenadas = [
    "BusinessTravel",
    "Department",
    "EducationField",
    "Gender",
    "JobRole",
    "MaritalStatus",
    "OverTime",
]

colunas_categoricas_ordenadas = [
    "Education",
    "EnvironmentSatisfaction",
    "JobInvolvement",
    "JobSatisfaction",
    "JobLevel",
    "PerformanceRating",
    "RelationshipSatisfaction",
    "StockOptionLevel",
    "WorkLifeBalance",
]

colunas_alvos = [
    "Attrition"
]

colunas_numericas = [
    coluna for coluna in df_colunas_removidas.columns \
        if coluna not in (colunas_categoricas_nao_ordenadas + colunas_categoricas_ordenadas + colunas_alvos)
    ]

display(df_colunas_removidas[colunas_numericas])
```

```python
# Listagem de todos as colunas categoricas
colunas_categoricas_totais = colunas_categoricas_nao_ordenadas + colunas_categoricas_ordenadas

# Identificando o maior tamanho no nome da coluna
contador = {coluna: len(coluna) for coluna in colunas_categoricas_totais}

for coluna in colunas_categoricas_totais:
    print(f"{coluna:<{max(contador.values())}}: {df_colunas_removidas[coluna].unique()}")
```

```python
df_colunas_removidas.info()
# Identificando o tamanho da base de dados
```

```python
# transformar os tipos não ordenados para categoria

for coluna in colunas_categoricas_nao_ordenadas:
    df_colunas_removidas[coluna] = df_colunas_removidas[coluna].astype("category")
```

```python
# Reduzindo o tamanho da base de dados apó a mudança dos itens para categoria
df_colunas_removidas.info()
```

```python
for coluna in df_colunas_removidas.select_dtypes("category"):
    print(f"{coluna}")
    print(f"Categories: {df_colunas_removidas[coluna].cat.categories}")
    print(f"Ordered: {df_colunas_removidas[coluna].cat.ordered}")
    print(f"Codes: {df_colunas_removidas[coluna].cat.codes.unique()}")
    print()
```

```python
display(df_colunas_removidas["Department"].value_counts().plot(kind="bar"))
```

```python
# Agrupamentos: contagem de funcionários por departamento
df_contagem_funcionarios_por_departamento = df_colunas_removidas.pivot_table(
    index="Department", columns="Attrition", aggfunc="size", observed=False)
df_contagem_funcionarios_por_departamento.insert(
        len(df_contagem_funcionarios_por_departamento.columns), 
        "Total", 
        df_contagem_funcionarios_por_departamento["No"] + df_contagem_funcionarios_por_departamento["Yes"]
    )
df_contagem_funcionarios_por_departamento.loc["Total"] = df_contagem_funcionarios_por_departamento.sum(0).values
display(df_contagem_funcionarios_por_departamento)

df_contagem_funcionarios_por_departamento.drop(columns="Total")[:-1].plot(kind="bar")
```

```python
# Agrupamentos: média de salário por departamento
df_salario_medio_por_departamento = df_colunas_removidas.pivot_table(index="Department", columns="Attrition", aggfunc="mean", values="MonthlyIncome")
with pd.option_context("float_format", "$ {:_.2f}".format):
    display(df_salario_medio_por_departamento)
```

```python
# Agrupamentos: gênero por departamento
df_contagem_genero_por_departamento = df_colunas_removidas.pivot_table(index="Department", columns=["Attrition", "Gender"], aggfunc="size")
display(df_contagem_genero_por_departamento)
```

```python
# Agrupamentos: gênero por cargo
df_contagem_genero_por_cargo = df_colunas_removidas.pivot_table(index="JobRole", columns=["Attrition", "Gender"], aggfunc="size")
display(df_contagem_genero_por_cargo)
```

```python
# Agrupamentos: média de salário por cargo
df_salario_medio_por_cargo = df_colunas_removidas.pivot_table(index="JobRole", columns="Attrition", aggfunc="mean", values="MonthlyIncome")
with pd.option_context("float_format", "$ {:_.2f}".format):
    display(df_salario_medio_por_cargo)
```

```python
# Agrupamentos: média de salário por cargo e nível
df_salario_medio_por_cargo_e_nivel = df_colunas_removidas.pivot_table(index="JobRole", columns=["Attrition", "JobLevel"], aggfunc="mean", values="MonthlyIncome").fillna("")
with pd.option_context("float_format", "$ {:_.2f}".format):
    display(df_salario_medio_por_cargo_e_nivel)
```

```python
# Agrupamentos: gênero por departamento
df_contagem_genero_por_departamento = df_colunas_removidas.pivot_table(index="Department", columns=["Attrition", "Gender"], aggfunc="size")
display(df_contagem_genero_por_departamento)
```

```python
# Agrupamentos: média de salário por nível educacional
df_salario_medio_por_nivel_educacional = df_colunas_removidas.pivot_table(index="Education", columns="Attrition", aggfunc="mean", values="MonthlyIncome")
with pd.option_context("float_format", "$ {:_.2f}".format):
    display(df_salario_medio_por_nivel_educacional)
```
