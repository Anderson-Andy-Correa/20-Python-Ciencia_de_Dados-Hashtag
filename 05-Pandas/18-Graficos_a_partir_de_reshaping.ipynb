# Pandas básico

```python
import pandas as pd

BASE = r"./dados/cereal_sem_negativos.csv"

df = pd.read_csv(BASE)

df.head()
```

```python
df.info()
```

```python
df.groupby(["mfr", "shelf"]).count()
```

```python
df.groupby(["mfr", "shelf"]).size().to_frame()
```

```python
df.groupby(["mfr", "shelf"]).size().to_frame().plot(kind="bar")
```

```python
df.groupby(["mfr", "shelf"]).size().unstack("shelf").fillna("-")
```

```python
plot = df.groupby(["mfr", "shelf"]).size().unstack("shelf") \
    .fillna(0).plot(
        kind="bar",
        rot=True,
        title="Quantidade de cereais por prateleira",
        # stacked=True, 
        figsize=(10, 5),
        xlabel="Fabricante",
        ylabel="Quantidade",
        )
for container in plot.containers:
    plot.bar_label(container)
plot.legend(title="Prateleira")
```

```python
df.groupby(["mfr", "shelf"]).size().unstack().stack()
```

```python
df.pivot_table(
    index="mfr",
    columns="shelf",
    aggfunc="size"
)
```

```python
df.pivot_table(
    index="mfr",
    columns="shelf",
    aggfunc="size"
).plot(kind="bar",
    rot=True,
    title="Quantidade de cereais por pratileira",
    stacked=True, 
    figsize=(10, 5),
    xlabel="Fabricante",
    ylabel="Quantidade",
)
```

```python
table = df.pivot_table(
    index="mfr",
    columns="type",
    aggfunc="size"
)
plot = table.plot(kind="bar",
    rot=True,
    title="Quantidade de cereais por tipo",
    figsize=(10, 5),
    xlabel="Fabricante",
    ylabel="Quantidade",
)

#labels = {label for label in plot}
plot.legend(title="Tipo")
```

```python
df.pivot_table(
    index="mfr",
    columns="vitamins",
    aggfunc="size"
).plot(kind="bar",
    rot=True,
    title="Quantidade de cereais por pratileira",
    figsize=(10, 5),
    xlabel="Fabricante",
    ylabel="Quantidade",
)
```
